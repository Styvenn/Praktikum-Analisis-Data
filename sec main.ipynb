```python
%pip install pandas seaborn matplotlib
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Using pandas sample dataset
data = pd.read_csv('nilai_siswa.csv')
```

```python
data.info()
```

```python
data.head()
```

```python
data.describe()
```

```python
print("Rata-rata:", data['Nilai'].mean())
print("Median:", data['Nilai'].median())
print("Modus:", data['Nilai'].mode()[0])
```

```python
matematika = data[data['Matpel'] == 'Matematika']
print(matematika)
```

```python
Inggris = data[data['Matpel'] == 'Bahasa Inggris']
print(Inggris)
```

```python
Fisika = data[data['Matpel'] == 'Fisika']
print(Fisika)
```

```python
Produktif = data[data['Matpel'] == 'Produktif']
print(Produktif)
```

```python
data.groupby('Matpel')['Nilai'].agg(['max','min'])
```

```python
rata = data.groupby('Matpel')['Nilai'].mean()
rata.plot(kind='bar')
plt.title('Rata-Rata Nilai per Matpel')
plt.xlabel('Mata Pelajaran')
plt.ylabel('Nilai Rata-Rata')
plt.show()
```

```python
sns.boxplot(x='Matpel', y='Nilai', data=data)
plt.title('Sebaran Nilai per Mata Pelajaran')
plt.show()
```

Nilai terendah ada pada mapel Bahasa Indonesia.
